```python
# Clear variables / output
from IPython.display import clear_output
clear_output(wait=True)

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt


from ISLP import load_data

# Load Auto dataset
Auto = load_data("Auto")

display(Auto.head())
Auto.info()
Auto.describe()
```

```python
plt.figure(figsize=(10, 6))

Auto = Auto[(Auto["origin"] == 1) & (Auto["year"] == 75)]

plt.scatter(Auto["weight"], Auto["mpg"])

for name, row in Auto.iterrows():
    plt.text(
        row["weight"],
        row["mpg"],
        name,
        fontsize=6
    )

plt.xlabel("Weight")
plt.ylabel("MPG")
plt.show()
```

```python
from sklearn.linear_model import LinearRegression

X = Auto[["weight"]]
y = Auto["mpg"]

fit = LinearRegression()
fit.fit(X, y)

x = np.linspace(Auto["weight"].min(), Auto["weight"].max(), 100)

plt.figure(figsize=(10, 6))
plt.scatter(Auto["weight"], Auto["mpg"])

# Add car names
for name, row in Auto.iterrows():
    plt.text(row["weight"], row["mpg"], name, fontsize=6)

# Add regression line
plt.plot(
    x,
    fit.predict(x.reshape(-1, 1)),
    color="red",
    linewidth=2
)

plt.xlabel("Weight")
plt.ylabel("MPG")
plt.show()

print("Intercept:", fit.intercept_)
print("Slope:", fit.coef_[0])
```

```python
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import PolynomialFeatures, StandardScaler

fit_poly = make_pipeline(
    StandardScaler(),
    PolynomialFeatures(degree=10),
    LinearRegression()
)

fit_poly.fit(X, y)

plt.figure(figsize=(10, 6))

plt.scatter(Auto["weight"], Auto["mpg"])

for name, row in Auto.iterrows():
    plt.text(
        row["weight"],
        row["mpg"],
        name,
        fontsize=6
    )

plt.plot(
    x,
    fit_poly.predict(x.reshape(-1, 1)),
    color="blue",
    linewidth=2
)

plt.xlabel("Weight")
plt.ylabel("MPG")
plt.show()
```

```python

from sklearn.cluster import KMeans

X_cluster = Auto[["weight", "mpg"]]

kmeans_result = KMeans(
    n_clusters=2,
    random_state=1,
    n_init=10
)

clusters = kmeans_result.fit_predict(X_cluster)

plt.figure(figsize=(10, 6))

plt.scatter(
    Auto["weight"],
    Auto["mpg"],
    c=clusters
)

for name, row in Auto.iterrows():
    plt.text(
        row["weight"],
        row["mpg"],
        name,
        fontsize=6
    )

plt.xlabel("Weight")
plt.ylabel("MPG")
plt.show()
```
